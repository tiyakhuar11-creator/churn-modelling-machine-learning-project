```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

y_test = Actual (real) answers from the dataset.
y_pred = Predicted answers generated by the model.

We store the predictions in a variable named y_pred.

```python
from sklearn.metrics import accuracy_score 
```

accuracy is use in classification data.in logicalregression that give predction on yes or no.

```python
round(accuracy_score(y_pred,y_test),2)*100
```
